```python
def preparar_features(df):

    datos = df.copy()

    # -------------------------------------------
    # 1. Número del mes
    # -------------------------------------------

    mapa_meses = {
        "Enero": 1,
        "Febrero": 2,
        "Marzo": 3,
        "Abril": 4,
        "Mayo": 5,
        "Junio": 6,
        "Julio": 7,
        "Agosto": 8,
        "Septiembre": 9,
        "Octubre": 10,
        "Noviembre": 11,
        "Diciembre": 12
    }

    datos["numero_mes"] = datos["Meses"].map(mapa_meses)

    # -------------------------------------------
    # 2. Crear fecha
    # -------------------------------------------

    datos["fecha"] = pd.to_datetime(
        dict(
            year=datos["Año"],
            month=datos["numero_mes"],
            day=1
        )
    )

    # -------------------------------------------
    # 3. Trimestre
    # -------------------------------------------

    datos["trimestre"] = datos["fecha"].dt.quarter

    # -------------------------------------------
    # 4. Nivel de ocupación
    # -------------------------------------------

    datos["nivel_ocupacion"] = pd.cut(
        datos["ocupacion"],
        bins=[0, 40, 60, 100],
        labels=["Baja", "Media", "Alta"],
        right=False
    )

    # -------------------------------------------
    # 5. Temporada alta
    # -------------------------------------------

    meses_alta = [
        "Enero",
        "Junio",
        "Julio",
        "Noviembre",
        "Diciembre"
    ]

    datos["temporada_alta"] = (
        datos["Meses"]
        .isin(meses_alta)
        .astype(int)
    )

    # -------------------------------------------
    # 6. Desviación respecto a media regional
    # -------------------------------------------

    media_region = (
        datos
        .groupby("region")["ocupacion"]
        .transform("mean")
    )

    datos["desviacion_media_region"] = (
        datos["ocupacion"] - media_region
    )

    # -------------------------------------------
    # 7. Escalado Min-Max
    # -------------------------------------------

    minimo = datos["ocupacion"].min()
    maximo = datos["ocupacion"].max()

    datos["ocupacion_escalada"] = (
        (datos["ocupacion"] - minimo)
        /
        (maximo - minimo)
    )

    # -------------------------------------------
    # 8. One-Hot Encoding
    # -------------------------------------------

    datos = pd.get_dummies(
        datos,
        columns=["region", "Meses"],
        prefix=["region", "mes"],
        dtype=int
    )

    return datos
```

## Cierre orientado a los objetivos del proyecto

Esta sección contrasta los hallazgos con los objetivos de negocio y las guías del curso. Las cifras se recalculan desde las fuentes del repositorio al ejecutar la celda siguiente. El backtest es temporal (entrena con el año previo y evalúa meses futuros); no representa una estimación causal del terremoto.

La libreta ya incluye preparación y limpieza, análisis descriptivo y visualizaciones, variables derivadas, codificación/escalado reproducibles, K-Means, agrupamiento jerárquico, DBSCAN y clasificación con conjunto de prueba, matriz de confusión, precisión, recall y F1. Apriori no aplica a las fuentes disponibles, pues no hay transacciones de compra. La celda registra y trata dos incidencias de calidad: cinco ocupaciones faltantes se mantienen como nulas (no como cero) y la capacidad RNT repetida por mes se agrega usando el dato más reciente por región. Las validaciones muestran cero claves duplicadas y cero ocupaciones fuera del rango 0-100%; no se eliminan valores válidos por ser extremos sin evidencia de error. La celda final de auditoria resume la cobertura, conciliacion de bases y metricas de prueba para facilitar la revision.

```python
from pathlib import Path
import unicodedata
import numpy as np
import pandas as pd

# Encuentra la raiz aunque Jupyter se inicie desde notebooks/.
raiz = next(
    (p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
     if (p / "Turismov1.2/Turismov1/data/02_turismo_colombia_limpio.xlsx").exists()),
    None,
)
if raiz is None:
    raise FileNotFoundError("No se encontro la carpeta de datos del proyecto desde el directorio actual.")
data_dir = raiz / "Turismov1.2" / "Turismov1" / "data"


def normalizar_columna(valor):
    texto = unicodedata.normalize("NFKD", str(valor))
    texto = "".join(c for c in texto if not unicodedata.combining(c))
    return texto.strip().lower().replace(" ", "_")


def cargar_hoja_ema(nombre, encabezado):
    tabla = pd.read_excel(data_dir / "anex-EMA-jul2026 (1).xlsx", sheet_name=nombre, header=encabezado)
    tabla.columns = [normalizar_columna(c) for c in tabla.columns]
    return tabla


ocupacion = pd.read_excel(data_dir / "02_turismo_colombia_limpio.xlsx")
ocupacion.columns = [normalizar_columna(c) for c in ocupacion.columns]
meses_orden = ["enero", "febrero", "marzo", "abril", "mayo", "junio",
               "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"]
ocupacion["meses"] = ocupacion["meses"].astype(str).str.strip().str.lower()
ocupacion["fecha_mes"] = pd.to_datetime(
    ocupacion["ano"].astype(int).astype(str) + "-" +
    ocupacion["meses"].map({m: i + 1 for i, m in enumerate(meses_orden)}).astype(str) + "-01"
)

print(f"Dataset: {len(ocupacion):,} filas; {ocupacion['region'].nunique()} regiones; "
      f"{ocupacion['ocupacion'].notna().sum():,} ocupaciones disponibles; "
      f"{ocupacion['ocupacion'].isna().sum()} faltantes.")
print("Faltantes por region/mes:")
print(ocupacion.loc[ocupacion["ocupacion"].isna(), ["region", "fecha_mes"]].to_string(index=False))
claves = ["ano", "meses", "region"]
print(f"Duplicados de clave (ano, mes, region): {ocupacion.duplicated(claves).sum()}")
fuera_rango = ocupacion["ocupacion"].notna() & ~ocupacion["ocupacion"].between(0, 100)
print(f"Ocupaciones fuera del rango valido 0-100%: {fuera_rango.sum()}")
# Los nulos no se imputan como cero; pandas los excluye de promedios y errores.

media_eje = ocupacion.loc[ocupacion["region"].eq("Eje Cafetero"), "ocupacion"].mean()
por_mes = (ocupacion.loc[ocupacion["region"].eq("Eje Cafetero")]
           .groupby("meses", sort=False)["ocupacion"].mean().reindex(meses_orden).dropna())
print(f"\nOcupacion media Eje Cafetero: {media_eje:.2f}%")
print("Tres meses de menor ocupacion media regional:")
print(por_mes.sort_values().head(3).map(lambda x: f"{x:.2f}%").to_string())

# Cada mes de ago.-oct. 2025 se compara con el mismo mes de 2024.
ocupacion["anio"] = ocupacion["fecha_mes"].dt.year
ocupacion["mes_num"] = ocupacion["fecha_mes"].dt.month
base_2024 = ocupacion.loc[ocupacion["anio"].eq(2024), ["region", "mes_num", "ocupacion"]]
prueba_2025 = ocupacion.loc[
    ocupacion["anio"].eq(2025) & ocupacion["mes_num"].between(8, 10),
    ["region", "mes_num", "ocupacion"],
]
backtest = prueba_2025.merge(base_2024, on=["region", "mes_num"], suffixes=("_real", "_pronostico"))
backtest = backtest.dropna(subset=["ocupacion_real", "ocupacion_pronostico"])
backtest = backtest.loc[backtest["ocupacion_real"].ne(0)]
mape = (backtest["ocupacion_real"].sub(backtest["ocupacion_pronostico"]).abs()
        .div(backtest["ocupacion_real"].abs()).mean() * 100)
print(f"\nBacktest estacional ago.-oct. 2025: MAPE {mape:.2f}% ({len(backtest)} pares region-mes).")

base_eje = ocupacion.loc[
    ocupacion["region"].eq("Eje Cafetero") & ocupacion["anio"].eq(2025)
    & ocupacion["mes_num"].between(8, 10), ["mes_num", "ocupacion"]
].set_index("mes_num")["ocupacion"]
pronostico_eje = base_eje.copy()
pronostico_eje.index = pronostico_eje.index.map({8: "Agosto", 9: "Septiembre", 10: "Octubre"})
print("Pronostico estacional Eje Cafetero 2026 (referencia, no efecto causal):")
print(pronostico_eje.map(lambda x: f"{x:.2f}%").to_string())

# La hoja contiene variaciones anuales, no niveles de ingresos.
ingresos = cargar_hoja_ema("1.1 V.A Ing.real", encabezado=12)
ingresos["ano"] = pd.to_numeric(ingresos["ano"].astype(str).str.extract(r"(\d{4})")[0], errors="coerce").ffill()
ingresos["meses"] = ingresos["meses"].astype(str).str.strip().str.lower()
ingresos_julio = ingresos.loc[ingresos["meses"].eq("julio")].set_index("ano")
columnas_regionales = [c for c in ingresos.columns[3:15] if c in ingresos_julio.columns]
crecimiento = {}
for region in columnas_regionales:
    cambios = pd.to_numeric(ingresos_julio.reindex(range(2020, 2027))[region], errors="coerce")
    if len(cambios) == 7 and cambios.notna().all() and (cambios > -100).all():
        crecimiento[region.strip()] = ((1 + cambios / 100).prod() - 1) * 100
ranking_ingresos = pd.Series(crecimiento, name="cambio_relativo_pct").sort_values(ascending=False)
print("\nCambio relativo encadenado de ingresos reales, julio 2019-julio 2026:")
print(ranking_ingresos.map(lambda x: f"{x:+.2f}%").to_string())
print("San Andres: no comparable; julio de 2020 fue -100% y julio de 2021 no tiene dato.")
if "eje cafetero" in ranking_ingresos.index:
    posicion = ranking_ingresos.index.get_loc("eje cafetero") + 1
    print(f"\nEje Cafetero: puesto {posicion} de {len(ranking_ingresos)} regiones comparables.")
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score

archivos_auditoria = {
    "01 integrado": "01_turismo_colombia_integrado.xlsx",
    "02 limpio": "02_turismo_colombia_limpio.xlsx",
    "03 modelado": "03_turismo_colombia_modelado.xlsx",
    "04 segmentado": "04_turismo_colombia_segmentado.xlsx",
    "05 prueba": "05_turismo_colombia_prediccion.xlsx",
}

bases_auditoria = {
    nombre: pd.read_excel(data_dir / archivo)
    for nombre, archivo in archivos_auditoria.items()
}
print("Filas por etapa:")
for nombre, base in bases_auditoria.items():
    print(f"{nombre}: {len(base):,}")

integrado = bases_auditoria["01 integrado"]
limpio = bases_auditoria["02 limpio"]
mascara_mensual = integrado["Meses"].astype(str).str.strip().str.lower().isin(meses_orden)
claves = ["A\u00f1o", "Meses", "region"]
comparacion = integrado.loc[mascara_mensual, claves + ["ocupacion"]].merge(
    limpio[claves + ["ocupacion"]],
    on=claves,
    how="outer",
    suffixes=("_integrado", "_limpio"),
    indicator=True,
)
coinciden_claves = comparacion["_merge"].eq("both")
valores_iguales = np.isclose(
    pd.to_numeric(comparacion.loc[coinciden_claves, "ocupacion_integrado"], errors="coerce").to_numpy(dtype=float),
    pd.to_numeric(comparacion.loc[coinciden_claves, "ocupacion_limpio"], errors="coerce").to_numpy(dtype=float),
    equal_nan=True,
)
print(f"Claves mensuales conciliadas: {coinciden_claves.sum():,}/{len(comparacion):,}")
print(f"Ocupaciones iguales (incluye nulos coincidentes): {valores_iguales.sum():,}/{len(valores_iguales):,}")
print(f"Ocupacion disponible/faltante en 02: {limpio['ocupacion'].notna().sum():,}/{limpio['ocupacion'].isna().sum():,}")

segmentado = bases_auditoria["04 segmentado"]
if "cluster_dbscan" in segmentado.columns:
    print(f"Registros DBSCAN marcados como ruido: {(segmentado['cluster_dbscan'] == -1).sum():,}")

prueba = bases_auditoria["05 prueba"]
real = prueba["real"].astype(int)
prediccion = prueba["prediccion"].astype(int)
tn, fp, fn, tp = confusion_matrix(real, prediccion, labels=[0, 1]).ravel()
print(f"Matriz TN/FP/FN/TP: {tn}/{fp}/{fn}/{tp}")
print(f"Metricas de prueba: accuracy={accuracy_score(real, prediccion):.2%}, "
      f"precision={precision_score(real, prediccion):.2%}, "
      f"recall={recall_score(real, prediccion):.2%}, "
      f"F1={f1_score(real, prediccion):.2%}")
```
